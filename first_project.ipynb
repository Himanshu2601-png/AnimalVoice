# My first ML project


## load

```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/dataprofessor/data/refs/heads/master/delaney_solubility_with_descriptors.csv')
df
```

##**data prepration**

### Data sepration using X and Y

```python
y=df['logS']
y
```

```python
X=df.drop('logS' ,axis =1)
X
```

### Data Splitting


```python
from sklearn.model_selection import train_test_split
X_train , X_test , y_train , y_test = train_test_split(X,y , test_size=0.2 , random_state=100)
```

```python
X_train
```

```python
X_test
```

# **Model Building**

### **Linear Regression**

### Training Model

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X_train , y_train)
```

## **Aplying the model for predxtion**

```python
y_lr_train_pred = lr.predict(X_train)
y_lr_test__pred = lr.predict(X_test)
```

```python
y_lr_train_pred
```

```python
y_lr_test__pred
```

**Model Performance Evaluatong Accuracy**

```python
y_train
```

```python
y_lr_train_pred
```

**Model performance**

```python
from sklearn.metrics import mean_squared_error ,r2_score
lr_train_men = mean_squared_error(y_train , y_lr_train_pred)
lr_train_r2 = r2_score(y_train , y_lr_train_pred)


lr_test_men = mean_squared_error(y_test , y_lr_test__pred)
lr_test_r2 = r2_score(y_test , y_lr_test__pred)

```

```python
print("LR MSE(train) :" , lr_train_men)
print("LR error2(train) :" , lr_train_r2)
print("LR MSE(test) :" , lr_test_men)
print("LR error2(test) :" , lr_test_r2)
```

```python
lr_result = pd.DataFrame([['Linear Regression', lr_train_men, lr_train_r2, lr_test_men, lr_test_r2]])
lr_result.columns = ["Method", "Training MSE", "Training R2", "Testing MSE", "Testing R2"]
```

```python
lr_result
```

# Random Forest

### **Traning Model**

```python
from numpy.random import RandomState
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor(max_depth= 2, random_state=100)
rf.fit(X_train , y_train)

```

## Applyting Model for Predictions

```python
y_rf_train_pred = rf.predict(X_train)
y_rf_test__pred = rf.predict(X_test)
```

### Evaluate Model performance

```python
from sklearn.metrics import mean_squared_error ,r2_score
rf_train_men = mean_squared_error(y_train , y_rf_train_pred)
rf_train_r2 = r2_score(y_train , y_rf_train_pred)


rf_test_men = mean_squared_error(y_test , y_rf_test__pred)
rf_test_r2 = r2_score(y_test , y_rf_test__pred)
```

```python
rf_result = pd.DataFrame([['Random Forest', rf_train_men, rf_train_r2, rf_test_men, rf_test_r2]])
rf_result.columns = ["Method", "Training MSE", "Training R2", "Testing MSE", "Testing R2"]
```

```python
rf_result
```

## two model comparison
